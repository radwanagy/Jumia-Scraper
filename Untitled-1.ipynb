```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
import csv


# إعداد الخيارات
options = Options()
options.add_argument("--start-maximized")  # يبدأ المتصفح بوضع ملئ الشاشة

# إعداد الخدمة بمسار chromedriver
service = Service(r"C:\Users\radwa\Downloads\chromedriver-win64\chromedriver-win64\chromedriver.exe")

# إنشاء المتصفح
driver = webdriver.Chrome(service=service, options=options)


# إنهاء الجلسة
#driver.quit()
```

```python
driver.get("https://www.jumia.com.eg/catalog/?q=samsung+a16")
```

```python
cards=driver.find_elements("xpath","//a[@ class='core']")
```

```python
with open("prodcts_of_jumia.csv",mode="w",newline="",encoding="utf-8") as file:
    writer = csv.writer(file)
    writer.writerow(["Title", "Image URL", "Price", "Rating"])
    for card in cards :
        try:
          title=driver.find_element("xpath", ".//h3[@class='name']").text
        except:
          title="Not Found"  
        try:
          price=driver.find_element("xpath", ".//div[@class='prc']").text
        except:
          price="Not Found" 
        try:
          image_url=driver.find_element("xpath", ".//img[@class='img']").get_attribute("src")
        except:
          image_url="Not Found"
        try:
          Rating=driver.find_element("xpath", ".//div[@class='n']").text
        except:
          Rating="Not Found"    
    
        writer.writerow([title, image_url, price, Rating])

driver.quit()                   
```
